### import read, visualize video and webcame

```python
path = cv2.VideoCapture('video_s.mp4')
```

```python
#does not work in colab
video = cv2.VideoCapture(path)
ret = True
while ret:
    ret, frame = video.read()
    if ret:
        cv2.imshow('frame', frame)
        cv2.waitKey(40) # 40 millisecond 25fps -> 1/25
```

### webcam

```python
# does not work in colab
web = cv2.VideoCapture(2)
while True:
    ret,frame = web.read()
    cv2_imshow(frame)
    if cv2.WaitKey(40) & 0xFF == ord('q'):
        break
web.release()
cv2.destroyAllWindows()
```
